```python
import torch
import matplotlib.pyplot as plt

plt.rcParams['figure.dpi'] = 120
plt.rcParams['axes.unicode_minus'] = True

# 正态分布概率密度函数
def normal_pdf(x: torch.Tensor, mu, sigma):
    return torch.exp(-(x - mu) ** 2 / (2 * sigma ** 2)) / (sigma * torch.sqrt(torch.tensor(2 * torch.pi)))

# 横坐标范围
x = torch.linspace(-300, 300, 1000)

mu = 100
settings = [
    {"sigma": 20, "label": "σ=20"},
    {"sigma": 60, "label": "σ=60"},
    {"sigma": 100, "label": "σ=100"},
]

for cfg in settings:
    y = normal_pdf(x, mu, cfg["sigma"])
    plt.plot(x.numpy(), y.numpy(), label=cfg["label"])

plt.xlabel("x")
plt.ylabel("概率密度")
plt.title(f"正态分布 N({mu}, σ²)")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```
