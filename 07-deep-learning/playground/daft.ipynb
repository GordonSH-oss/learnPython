```python
import torch
from torch import nn

# 直接实例化，不需要写class
model_seq = nn.Sequential(
    nn.Linear(2, 64),
    nn.ReLU(),
    nn.Linear(64, 32),
    nn.ReLU(),
    nn.Linear(32, 3)
)

# 前向传播，直接喂数据，内部自动一层一层运算
x = torch.randn(16, 2)   # batch=16，每个样本2维
out = model_seq(x)
print(out.shape) # torch.Size([16, 3])
```
